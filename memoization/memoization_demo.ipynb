```python
#fibonacci

import time
def  fib(n):
    if n==0 or n==1:
        return 1
    else:
        return fib(n-1)+fib(n-2)
```

```python
start = time.time()
fib(38)
print(time.time()-start)


# it takes a log time to evaluate,, 
# to solve this problem ,use memoization trick
```

```python
#fibonacci

import time
def  fib(n,d):
    if n in d:
        return d[n]
    else:
        d[n]= fib(n-1,d)+fib(n-2,d)
        return d[n]
```

```python
start = time.time()
d={0:1,1:1}
fib(380,d)
print(time.time()-start)
```
