```python
import pandas as pd

or_data = pd.DataFrame()

or_data["input1"] = [1, 1, 0, 0]
or_data["input2"] = [1, 0, 1, 0]
or_data["output"] = [1, 1, 1, 0]

print(or_data)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(
    x=or_data["input1"],
    y=or_data["input2"],
    hue=or_data["output"],
    s=200
)

plt.title("OR Gate")
plt.show()
```

```python
and_data = pd.DataFrame()

and_data["input1"] = [1, 1, 0, 0]
and_data["input2"] = [1, 0, 1, 0]
and_data["output"] = [1, 0, 0, 0]

print(and_data)
```

```python
sns.scatterplot(
    x=and_data["input1"],
    y=and_data["input2"],
    hue=and_data["output"],
    s=200
)

plt.title("AND Gate")
plt.show()
```

```python
xor_data = pd.DataFrame()

xor_data["input1"] = [1, 1, 0, 0]
xor_data["input2"] = [1, 0, 1, 0]
xor_data["output"] = [0, 1, 1, 0]

print(xor_data)
```

```python
sns.scatterplot(
    x=xor_data["input1"],
    y=xor_data["input2"],
    hue=xor_data["output"],
    s=200
)

plt.title("XOR Gate")
plt.show()
```

```python
from sklearn.linear_model import Perceptron
```

```python
clf1 = Perceptron()

clf2 = Perceptron()

clf3 = Perceptron()
```

```python
clf1.fit(
    and_data.iloc[:, 0:2].values,
    and_data.iloc[:, -1].values
)

clf2.fit(
    or_data.iloc[:, 0:2].values,
    or_data.iloc[:, -1].values
)

clf3.fit(
    xor_data.iloc[:, 0:2].values,
    xor_data.iloc[:, -1].values
)
```

```python
clf1.coef_
```

```python
clf1.intercept_
```

```python
plt.plot(x, y)

sns.scatterplot(
    data=and_data,
    x="input1",
    y="input2",
    hue="output",   # Agar column ka naam 'ouput' hi hai
    s=200
)

plt.show()
```

```python
plt.plot(x, y)

sns.scatterplot(
    x=or_data["input1"],
    y=or_data["input2"],
    hue=or_data["output"],
    s=200
)

plt.show()
```

```python
from mlxtend.plotting import plot_decision_regions
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 6))

plot_decision_regions(
    X=xor_data.iloc[:, 0:2].values,
    y=xor_data.iloc[:, -1].values,
    clf=clf3,
    legend=2
)

plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.title("XOR Gate Decision Boundary")

plt.show()
```
