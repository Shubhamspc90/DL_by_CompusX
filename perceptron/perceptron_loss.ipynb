Ye poora code Perceptron Algorithm ko scratch se samajhne aur visualize karne ke liye likha gaya hai. Iska main goal hai dekhna ki Perceptron kaise decision boundary (line) learn karta hai.

```python
from sklearn.datasets import make_classification

import numpy as np
```

```python
# creating datasets

X, y = make_classification(
    n_samples=100,
    n_features=2,
    n_informative=1,
    n_redundant=0,
    n_classes=2,
    n_clusters_per_class=1,
    random_state=41,
    hypercube=False,
    class_sep=15
)

# Explanation
# n_samples=100 → 100 data points create karega.
# n_features=2 → Har sample me 2 input features honge.
# n_informative=1 → Sirf 1 feature actual information carry karega.
# n_redundant=0 → Koi duplicate/redundant feature nahi hoga.
# n_classes=2 → 2 classes (0 aur 1).
# n_clusters_per_class=1 → Har class ka sirf 1 cluster hoga.
# random_state=41 → Har baar same dataset generate hoga.
# hypercube=False → Dataset hypercube structure me generate nahi hoga.
# class_sep=15 → Dono classes ke beech bahut zyada distance hoga (easily separable).
```

```python
print(X.shape)
y.shape
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=y,
    cmap="winter",
    s=100
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Generated Classification Dataset")

plt.show()


# Explanation
# plt.figure(figsize=(10,6)) → Figure ka size set karta hai.
# X[:, 0] → Pehla feature (X-axis).
# X[:, 1] → Dusra feature (Y-axis).
# c=y → Class ke according color assign karega.
# cmap="winter" → Winter color map use karega.
# s=100 → Points ka size 100 pixels.
```

```python
# Ye function Perceptron Algorithm ko scratch (without scikit-learn) implement karta hai.
# Iska purpose hai dataset se learning karke ek decision boundary (line) find karna, jisse 2 classes ko alag kiya ja sake.

def perceptron(X,y):
    w1=w2=b=1
    lr=0.1
    
    for j in range(1000):
        for i in range(X.shape[0]):
            # check Condition
            z=w1*X[i][0]+w2*X[i][1]+b   # This computes the weighted sum. ,formula=>  z=w1​x1​+w2​x2​+b 
            
            # Update the weights only if the prediction is wrong.
            if z*y[i]<0: 
                w1=w1+lr*y[i]*X[i][0]   # update formula (w1​=w1​+ηyx1)​
                w2=w2+lr*y[i]*X[i][1]
                b=b+lr*y[i]
                
    return w1,w2,b
                     
```

```python
w1,w2,b=perceptron(X,y)
print(w1)
print(w2)
print(b)
```

```python
# for graph ploting 

m = -(w1/w2)
c = -(b/w2)

print(m,c)
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input +c  # formula (y=mx+C)

plt.figure(figsize=(10, 6))
plt.plot(x_input, y_input, color="red", linewidth=3)
plt.scatter(
    X[:, 0],  # Sabhi rows ka first feature. , Ye X-axis par jayega.
    X[:, 1],   # Sabhi rows ka second feature. ,  Ye Y-axis par jayega.
    c=y, # Class ke according color dega.
    cmap="winter",
    s=100  # Points ka size. , Jitna bada number, utna bada point.
)

plt.ylim(-3, 2) # Y-axis sirf  -3 to 2  tk hi dikhegi ,,Isse graph zoom ho jata hai

plt.show()

# example 
# y
# 0
# 1
# 1
# Class 0 → Blue
# Class 1 → Green
```

**Pura Flow**

```python
# Dataset
#    │
#    ▼
# Train Perceptron
#    │
#    ▼
# Get Weights
# (w1,w2,b)
#    │
#    ▼
# Find Slope (m)
# Find Intercept (c)
#    │
#    ▼
# Generate X Values
# (np.linspace)
#    │
#    ▼
# Calculate Y Values
# (y = mx + c)
#    │
#    ▼
# Draw Decision Boundary
# (Red Line)
#    │
#    ▼
# Plot Dataset
# (Scatter Plot)
#    │
#    ▼
# Final Graph
```
