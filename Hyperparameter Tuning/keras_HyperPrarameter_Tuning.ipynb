```python
import kerastuner as kt
```

```python
# for 1. How to select Appropriate optimizer.

def built_model(hp): 
    model = Sequential() 
    model.add(Dense(32, activation='relu', input_dim=8)) 
    model.add(Dense(1, activation='sigmoid')) 
     
    Optimizer = hp.Choice(
        'optimizer',
        values=['adam', 'sgd', 'rmsprop', 'adadelta']
    ) 
     
    model.compile(
        optimizer=Optimizer,
        loss='binary_crossentropy',
        metrics=['accuracy']
    ) 
     
    return model
```

```python
tuner = kt.RandomSearch(
    built_model,
    objective='val_accuracy',
    max_trials=5,
    directory='my_dir',
    project_name='optimizer_test',
    overwrite=True
)

tuner.search(
    X_train,
    y_train,
    epochs=5,
    validation_data=(X_test, y_test)
)
```

```python
tuner.results_summary()
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model = tuner.get_best_models(num_models=1)[0]

model.summary()
```

```python
model.fit(X_train,y_train,batch_size=32,epochs=100,initial_epoch=6,validation_data=(X_test,y_test))
```

```python
# 3. How to select no. of layers.

def  built_model(hp):
    model=Sequential()
    
    model.add(Dense(48,activation='relu',input_dim=8))
    
    for i in range (hp.Int('num_layers',min_value=1,max_value=10)):
        model.add(Dense(48,activation='relu'))
    model.add(Dense(1,activation='sigmoid'))
    
    model.compile(optimizer='rmsprop',loss='binary_crossentropy',metrics=['accuracy'])
    
    return model
```

```python
#for  4. All in all one model.

# hyperparameters together:
# number of neurons (units), optimizer, learning rate, and number of layers
```

1. Tune Units + Optimizer

```python
def built_model(hp):
    model = Sequential()

    # Number of neurons
    units = hp.Int(
        'units',
        min_value=8,
        max_value=128,
        step=8
    )

    model.add(
        Dense(
            units=units,
            activation='relu',
            input_dim=8
        )
    )

    model.add(Dense(1, activation='sigmoid'))

    # Optimizer
    optimizer = hp.Choice(
        'optimizer',
        values=['adam', 'sgd', 'rmsprop', 'adadelta']
    )

    model.compile(
        optimizer=optimizer,
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model
```

2. Create Tuner

```python
tuner = kt.RandomSearch(
    built_model,
    objective='val_accuracy',
    max_trials=10,
    directory='my_dir',
    project_name='all_hyperparameters',
    overwrite=True
)
```

3. Search

```python
tuner.search(
    X_train,
    y_train,
    epochs=5,
    validation_data=(X_test, y_test)
)
```

4. Get the best values

```python
best_hp = tuner.get_best_hyperparameters(num_trials=1)[0]

print("Best Units:", best_hp.get('units'))
print("Best Optimizer:", best_hp.get('optimizer'))
```
